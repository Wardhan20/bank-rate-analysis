# UK Interest Rates, 1975-2026

In this project I analyse 50 years of the Bank of England's Interest Rate to see how it has changed and what that means for people

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
raw = pd.read_csv("data/boe_bank_rate.csv")
print(raw.shape)
display(raw.head())
display(raw.tail())
raw.info()
```

```python
df = raw.rename(columns = {"Date Changed": "Date", "Rate": "Rate"})
```

Renamed column names to short and has no space, so it is easier to type in my code and less likely to cause typos

```python
df.info()
```

```python
df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"], format = "%d %b %y")
```

pandas read the data as text while loading the CSV file. I changed them into real dates by using pd.to_datetime. As text, the dates couldn't be sorted in time order or plotted on a timeline.

```python
df.info()
```

```python
df = df.sort_values(["Date","Rate"]).reset_index(drop=True)
```

```python
df
```

Sorted the oldest rows first because the original data starts with 2026 at the top. This solves the timelines and working out things like how much did the rate changed from the previous one. Used reset_index(drop=True) so the rows numbered in new order
